```python
u = np.array([2, 3, 4, 6])
u
```

Multiplicación

```python
u * v
```

```python
v.shape[0]
```

```python
def multiplicacion_de_vectores (u, v):
    assert u.shape[0] == v.shape[0]

    n = u.shape[0]

    result = 0.0

    for i in range (n):
        result = result + u[i] * v[i]

    return result
```

```python
multiplicacion_de_vectores (u, v)
```

```python
u.dot(v)
```

Multiplicación de vector por matriz

```python
U = np.array([
    [2, 4, 5, 6],
    [1, 2, 1, 2],
    [3, 1, 2, 1],
])
```

```python
U.shape[1]
```

```python
def vector_por_matriz (U, v):
    assert U.shape[1] == v.shape[0]

    num_rows = U.shape[0]

    result = np.zeros(num_rows)

    for i in range(num_rows):
        result[i] = multiplicacion_de_vectores(U[i], v)
    
    return result
```

```python
vector_por_matriz(U, v)
```

```python
U.dot(v)
```

Multiplicación de matriz por matriz

```python
V = np.array([
    [2, 4, 5],
    [1, 2, 1],
    [3, 1, 2],
    [6, 8, 9],
])
```

```python
def matriz_por_matriz(U, V):
    assert U.shape[1] == V.shape[0]

    num_rows = U.shape[0]
    num_cols = V.shape[1]

    result = np.zeros((num_rows, num_cols))

    for i in range(num_cols):
        vi = V[:, i]
        Uvi = vector_por_matriz(U, vi)
        result[:, i] = Uvi

    return result
```

```python
matriz_por_matriz(U, V)
```
